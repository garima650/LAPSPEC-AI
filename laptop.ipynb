```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2, random_state=42)  ## Here we are splitting the data into training and testing data. test_size=0.2 means 20% of the data will be used for testing and 80% will be used for training. random_state=42 means we are using a random seed of 42 to get the same result every time we run the code.
```

```python
from sklearn.svm import SVR
svr = SVR()
svr.fit(x_train , y_train)
```

```python
svr.score(x_test , y_test)  ## Here we are checking the accuracy of the model. The score method returns the coefficient of determination R^2 of the prediction. The best possible score is 1.0 and it can be negative (because the model can be arbitrarily worse). A constant model that always predicts the expected value of y, disregarding the input features, would get a R^2 score of 0.0.
```

```python
import joblib
import pandas as pd
model=joblib.load('laptop_price_predictor_model.lib')
  ## Here we are loading the model from the file named laptop_price_predictor_model.lib using joblib library.
le_company=joblib.load('company_encoder.lib')
le_op=joblib.load('op_encoder.lib')
le_cpu=joblib.load('cpu_encoder.lib')
le_gpu=joblib.load('gpu_encoder.lib')
le_memory=joblib.load('memory_encoder.lib')
le_type=joblib.load('type_encoder.lib')
le_screen=joblib.load('screen_resolution_encoder.lib')
```

```python
new_laptop = {
    'Company': 'HP',
    'TypeName': 'Notebook',
    'Inches': 15.6,
    'ScreenResolution': 'Full HD 1920x1080',
    'Cpu': 'Intel Core i5 7200U 2.5GHz',
    'Ram': 8,              # Already numeric
    'Memory': '256GB SSD', # Make sure this matches your final numeric column layout!
    'Gpu': 'Intel HD Graphics 620',
    'OpSys': 'Windows 10',
    'Weight': 1.86         # Already numeric
}
input_df=pd.DataFrame([new_laptop])  ## Here we are creating a new data frame from the new_laptop dictionary. The pd.DataFrame() function is used to create a data frame from a dictionary. The [new_laptop] is used to create a list of dictionaries. The list of dictionaries is used to create a data frame.
```

```python
input_df['Company'] = le_company.transform(input_df['Company'])
input_df['OpSys'] = le_op.transform(input_df['OpSys'])
input_df['Cpu'] = le_cpu.transform(input_df['Cpu'])
input_df['Gpu'] = le_gpu.transform(input_df['Gpu'])
input_df['Memory'] = le_memory.transform(input_df['Memory'])
input_df['TypeName'] = le_type.transform(input_df['TypeName'])
input_df['ScreenResolution'] = le_screen.transform(input_df['ScreenResolution'])
```

```python
# 1. Force the columns to match the exact order of your training dataset
feature_order = list(x_train.columns)
input_df = input_df[feature_order]

# 2. Now run the prediction (it will execute perfectly!)
predicted_price = model.predict(input_df)[0]
```
